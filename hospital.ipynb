```python
df = pd.read_csv('hospital_patient_records_dirty.csv') 
df.shape 
df.info() 
df.describe(include='all')
```

```python
print(df.info())
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Rows after removing duplicates:", len(df))
```

```python
text_cols = ['Name', 'Gender', 'Diagnosis', 'City', 'BloodType', 'HasInsurance']
for col in text_cols:
    df[col] = df[col].astype('string').str.strip()
```

```python
df['Gender'] = df['Gender'].str.lower()

df['Gender'] = df['Gender'].replace({
    'm': 'Male',
    'male': 'Male',
    'f': 'Female',
    'female': 'Female',
    '?': np.nan
})
```

```python
df['Diagnosis'] = df['Diagnosis'].str.lower()

df['Diagnosis'] = df['Diagnosis'].replace({
    'asthma': 'Asthma',
    'athsma': 'Asthma',

    'arthritis': 'Arthritis',

    'bronchitis': 'Bronchitis',

    'anemia': 'Anemia',
    'anaemia': 'Anemia',

    'migraine': 'Migraine',
    'migrane': 'Migraine',

    'flu': 'Influenza',
    'influenza': 'Influenza',

    'diabetes': 'Type 2 Diabetes',
    'diabetis': 'Type 2 Diabetes',
    'type 2 diabetes': 'Type 2 Diabetes',

    'hypertension': 'Hypertension',
    'hyper tension': 'Hypertension',
    'high blood pressure': 'Hypertension'
})
```

```python
df['City'] = df['City'].str.lower().str.strip()

df['City'] = df['City'].replace({
    'karachi': 'Karachi',
    'lahore': 'Lahore',
    'islamabad': 'Islamabad',
    'rawalpindi': 'Rawalpindi',
    'multan': 'Multan',
    'peshawar': 'Peshawar',
    'quetta': 'Quetta',
    'faisalabad': 'Faisalabad',
    'unknown': np.nan
})
```

```python
df['HasInsurance'] = df['HasInsurance'].str.lower()

df['HasInsurance'] = df['HasInsurance'].replace({
    'yes': 1,
    'y': 1,
    '1': 1,
    'no': 0,
    'n': 0,
    '0': 0,
    '--': np.nan
})
```

```python
df.loc[(df['Age'] < 0) | (df['Age'] > 100), 'Age'] = np.nan

df.loc[(df['HeightCM'] < 100) | (df['HeightCM'] > 250), 'HeightCM'] = np.nan

df.loc[(df['WeightKG'] < 20) | (df['WeightKG'] > 250), 'WeightKG'] = np.nan

df.loc[(df['SystolicBP'] < 60) | (df['SystolicBP'] > 250), 'SystolicBP'] = np.nan
```

```python
df['BillingAmount'] = (
    df['BillingAmount']
    .astype('string')
    .str.replace('$', '', regex=False)
    .str.replace('PKR', '', regex=False)
    .str.strip()
)

df['BillingAmount'] = pd.to_numeric(df['BillingAmount'], errors='coerce')
```

```python
df['AdmissionDate'] = pd.to_datetime(
    df['AdmissionDate'],
    errors='coerce',
    dayfirst=False
)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
numeric_cols = [
    'Age',
    'HeightCM',
    'WeightKG',
    'SystolicBP',
    'BillingAmount'
]

categorical_cols = [
    'Gender',
    'Diagnosis',
    'City',
    'BloodType',
    'HasInsurance'
]
```

```python
numeric_pipeline = SimpleImputer(strategy='median')

categorical_pipeline = SimpleImputer(
    strategy='most_frequent'
)
```

```python
preprocessor = ColumnTransformer([
    ('numeric', numeric_pipeline, numeric_cols),
    ('categorical', categorical_pipeline, categorical_cols)
])
```

```python
X = df[numeric_cols + categorical_cols]

X_clean = preprocessor.fit_transform(X)

print(X_clean.shape)
```

```python
from sklearn.preprocessing import StandardScaler

X_numeric = df[numeric_cols]

X_numeric = SimpleImputer(strategy='median').fit_transform(X_numeric)

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_numeric)

print(X_scaled[:5])
```

```python
df['AdmissionDate'] = df['AdmissionDate'].dt.strftime('%Y-%m-%d')

df.to_csv("hospital_patient_records_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
